## Paso 2 · Los datos de entrada
Una lista de solicitudes inventadas. Cada elemento es un texto.

```python
import pandas as pd

solicitudes = [
    "Me cobraron dos veces el servicio de agua.",
    "Hay un hueco grande en la calle de mi barrio.",
    "El alumbrado de la cancha lleva una semana dañado.",
    "Necesito la factura del mes pasado.",
    "Llevo tres días sin agua en la casa.",
    "El semáforo de la esquina no funciona.",
    "Me llegó un cobro que no entiendo.",
    "Hay una fuga de agua en el andén.",
]

tabla = pd.DataFrame({"solicitud": solicitudes})
tabla
```

## Paso 3 · Una función con condiciones
La función `clasificar` hace tres cosas, en este orden:
1. **Valida** que la entrada sea un texto que no esté vacío (si no, se detiene con un error).
2. **Cuenta** cuántas palabras clave de cada tema aparecen.
3. **Decide con una condición:** si ningún tema aparece → `otro`; si dos temas empatan → `revisión humana`; si no, el tema con más puntos.

🔮 **Antes de ejecutar el Paso 4**, mira la lista `PALABRAS` y escribe en el chat qué categoría esperas para: *«Me cobraron y además llevo tres días sin agua»*.

```python
PALABRAS = {
    "cobros": ["cobr", "factura", "pago"],
    "infraestructura": ["hueco", "calle", "alumbrado", "semáforo", "andén"],
    "servicio": ["sin agua", "sin luz", "fuga", "corte"],
}

def validar(texto):
    if not isinstance(texto, str):
        raise TypeError("La entrada debe ser un texto")
    if texto.strip() == "":
        raise ValueError("La entrada está vacía")
    return texto.strip().lower()

def clasificar(texto):
    texto = validar(texto)
    puntos = {tema: sum(p in texto for p in palabras) for tema, palabras in PALABRAS.items()}
    mejor = max(puntos.values())
    if mejor == 0:
        return "otro"
    ganadores = [tema for tema, n in puntos.items() if n == mejor]
    if len(ganadores) > 1:
        return "revisión humana"
    return ganadores[0]

def procesar(texto):
    """Lo que usaría la funcionaria: si algo falla, no inventa: pasa a revisión humana."""
    try:
        return clasificar(texto)
    except (TypeError, ValueError):
        return "revisión humana"

print("Funciones definidas: validar, clasificar, procesar")
```

## Paso 4 · Ejecuta sobre los datos
Compara con lo que escribiste en el chat.

```python
tabla["categoria"] = tabla["solicitud"].apply(procesar)
tabla
```

```python
print("Mensaje con dos temas a la vez:")
print(procesar("Me cobraron y además llevo tres días sin agua"))
```

## Paso 5 · Casos de aceptación
Tres casos que decides **antes** de mirar el resultado: dos normales y uno límite. Con ellos sabrás si tu archivo funciona, o si algo cambió sin que te dieras cuenta.

```python
CASOS = [
    ("Me cobraron dos veces el servicio de agua.", "cobros"),
    ("Hay un hueco grande en la calle.", "infraestructura"),
    ("", "revisión humana"),   # caso límite: entrada vacía
]

for texto, esperado in CASOS:
    obtenido = procesar(texto)
    marca = "✅" if obtenido == esperado else "❌"
    print(f"{marca} esperado: {esperado:<16} obtenido: {obtenido:<16} | {texto!r}")
```
